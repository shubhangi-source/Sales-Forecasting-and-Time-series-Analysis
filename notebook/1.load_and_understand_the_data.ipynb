# Import libraries

```python
import pandas as pd
import numpy as np
```

# Load dataset

```python
df = pd.read_csv("../data/train.csv")
```

# Random 5 rows


```python
print("\nRandom Sample:")
print(df.sample(5))
```

# Unique values in important columns

```python
print("\nUnique Stores:", df["Store"].nunique())
print("Unique Days:", df["DayOfWeek"].nunique())
```

# Check date range

```python
print("\nDate Range:")
print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())
```

# Shape and Summary of Dataset

```python
print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()
```

# Statistical summary

```python
print("\nStatistical Summary:")
print(df.describe())
```

# Missing values

```python
 
print("\nMissing Values:")
print(df.isnull().sum())
```

# Duplicate rows


```python

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

## Observation

- No missing values were found in the dataset.
- No duplicate records were found.
